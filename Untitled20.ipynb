```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
import random

# =========================
# CONFIG
# =========================
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
SEQ_LEN = 30
EPOCHS = 20
BATCH_SIZE = 64
RUNS = 3
MC_SAMPLES = 20
RUL_CAP = 125  # common practice for C-MAPSS

SEED = 42
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

# =========================
# LOAD DATA
# =========================
def load_cmapss(train_path, test_path, rul_path):
    cols = ['unit', 'cycle'] + [f'op{i}' for i in range(1,4)] + [f's{i}' for i in range(1,22)]

    train = pd.read_csv(train_path, sep='\s+', header=None)
    test = pd.read_csv(test_path, sep='\s+', header=None)
    rul = pd.read_csv(rul_path, header=None)

    train.columns = cols
    test.columns = cols

    # Compute RUL for train
    max_cycle = train.groupby('unit')['cycle'].max()
    train['RUL'] = train.apply(lambda row: max_cycle[row['unit']] - row['cycle'], axis=1)
    train['RUL'] = train['RUL'].clip(upper=RUL_CAP)

    return train, test, rul

# =========================
# PREPROCESS
# =========================
def preprocess(train):
    feature_cols = [c for c in train.columns if c.startswith('s')]
    scaler = MinMaxScaler()
    train[feature_cols] = scaler.fit_transform(train[feature_cols])
    return train, feature_cols, scaler

# =========================
# SEQUENCE CREATION
# =========================
def create_sequences(df, features):
    X, y = [], []
    for unit in df['unit'].unique():
        data = df[df['unit'] == unit]
        for i in range(len(data) - SEQ_LEN):
            X.append(data[features].iloc[i:i+SEQ_LEN].values)
            y.append(data['RUL'].iloc[i+SEQ_LEN])
    return np.array(X), np.array(y)

# =========================
# MODELS
# =========================
class LSTMModel(nn.Module):
    def __init__(self, input_size):
        super().__init__()
        self.lstm = nn.LSTM(input_size, 64, batch_first=True)
        self.dropout = nn.Dropout(0.3)
        self.fc = nn.Linear(64, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        out = out[:, -1, :]
        out = self.dropout(out)
        return self.fc(out)

class GRUModel(nn.Module):
    def __init__(self, input_size):
        super().__init__()
        self.gru = nn.GRU(input_size, 64, batch_first=True)
        self.dropout = nn.Dropout(0.3)
        self.fc = nn.Linear(64, 1)

    def forward(self, x):
        out, _ = self.gru(x)
        out = out[:, -1, :]
        out = self.dropout(out)
        return self.fc(out)

# =========================
# TRAINING
# =========================
def train_model(model, X, y):
    model.to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
    loss_fn = nn.MSELoss()

    X_t = torch.tensor(X, dtype=torch.float32).to(DEVICE)
    y_t = torch.tensor(y, dtype=torch.float32).to(DEVICE)

    for epoch in range(EPOCHS):
        model.train()
        optimizer.zero_grad()
        preds = model(X_t).squeeze()
        loss = loss_fn(preds, y_t)
        loss.backward()
        optimizer.step()
    return model

# =========================
# MC DROPOUT
# =========================
def predict_with_uncertainty(model, X, n_samples=MC_SAMPLES):
    model.train()  # keep dropout ON
    X_t = torch.tensor(X, dtype=torch.float32).to(DEVICE)

    preds = []
    for _ in range(n_samples):
        preds.append(model(X_t).detach().cpu().numpy())

    preds = np.array(preds)  # [samples, N, 1]
    mean = preds.mean(axis=0).squeeze()
    std = preds.std(axis=0).squeeze()
    return mean, std

# =========================
# BASELINE COST
# =========================
def baseline_cost(rul_pred, uncertainty):
    failure_risk = np.maximum(0, 50 - rul_pred)
    uncertainty_penalty = 10 * uncertainty
    return (failure_risk + uncertainty_penalty).mean()

# =========================
# CAMDE++ COMPONENTS
# =========================
def generate_context(n):
    # Example: [criticality, load]
    criticality = np.random.uniform(0.5, 1.5, n)
    load = np.random.uniform(0.5, 1.5, n)
    return np.stack([criticality, load], axis=1)

def camde_adjustment(rul_pred, uncertainty, context):
    criticality = context[:, 0]
    load = context[:, 1]

    # Φ: context-aware modulation
    adjusted_rul = rul_pred / (criticality * load)

    # risk term
    risk = uncertainty * criticality
    return adjusted_rul, risk

def camde_cost(adjusted_rul, risk):
    failure_cost = np.maximum(0, 40 - adjusted_rul)
    risk_penalty = 15 * risk
    return (failure_cost + risk_penalty).mean()

# =========================
# EXPERIMENT PIPELINE
# =========================
def run_experiment(train_path, test_path, rul_path):
    set_seed(SEED)

    train, test, rul = load_cmapss(train_path, test_path, rul_path)
    train, features, scaler = preprocess(train)
    X, y = create_sequences(train, features)

    results = {"LSTM": [], "GRU": [], "CAMDE++": []}

    for run in range(RUNS):
        print(f"Run {run+1}")

        for name, Model in [("LSTM", LSTMModel), ("GRU", GRUModel)]:
            set_seed(SEED + run)
            model = Model(len(features))
            model = train_model(model, X, y)

            preds, unc = predict_with_uncertainty(model, X)

            rmse = np.sqrt(mean_squared_error(y, preds))
            base_cost = baseline_cost(preds, unc)
            results[name].append((rmse, base_cost))

            # CAMDE++
            context = generate_context(len(preds))
            adj_rul, risk = camde_adjustment(preds, unc, context)
            c_cost = camde_cost(adj_rul, risk)

            results["CAMDE++"].append((rmse, c_cost))

    return results

# =========================
# SUMMARY
# =========================
def summarize(results, title):
    print(f"\n==== {title} ====")
    for model in results:
        rmses = [r[0] for r in results[model]]
        costs = [r[1] for r in results[model]]

        print(f"{model}: RMSE={np.mean(rmses):.3f}±{np.std(rmses):.3f}, "
              f"Cost={np.mean(costs):.3f}±{np.std(costs):.3f}")

# =========================
# RUN
# =========================
fd001 = run_experiment("train_FD001.txt", "test_FD001.txt", "RUL_FD001.txt")
fd004 = run_experiment("train_FD004.txt", "test_FD004.txt", "RUL_FD004.txt")

summarize(fd001, "FD001 RESULTS")
summarize(fd004, "FD004 RESULTS")
```

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
import random

# =========================
# CONFIG
# =========================
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
SEQ_LEN = 30
EPOCHS = 30
BATCH_SIZE = 64
RUNS = 3
MC_SAMPLES = 20
MAX_RUL = 125

# =========================
# SEED
# =========================
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

# =========================
# LOAD DATA
# =========================
def load_data(train_path, test_path, rul_path):
    cols = ['unit', 'cycle'] + [f'op{i}' for i in range(1,4)] + [f's{i}' for i in range(1,22)]

    train = pd.read_csv(train_path, sep='\s+', header=None)
    test = pd.read_csv(test_path, sep='\s+', header=None)
    rul = pd.read_csv(rul_path, header=None)

    train.columns = cols
    test.columns = cols

    # Train RUL
    max_cycle = train.groupby('unit')['cycle'].max()
    train['RUL'] = train.apply(lambda row: max_cycle[row['unit']] - row['cycle'], axis=1)
    train['RUL'] = train['RUL'].clip(upper=MAX_RUL) / MAX_RUL

    return train, test, rul

# =========================
# PREPROCESS
# =========================
def preprocess(train, test):
    features = [c for c in train.columns if c.startswith('s')]

    scaler = MinMaxScaler()
    train[features] = scaler.fit_transform(train[features])
    test[features] = scaler.transform(test[features])

    return train, test, features

# =========================
# SEQUENCES
# =========================
def create_train_sequences(df, features):
    X, y = [], []

    for unit in df['unit'].unique():
        data = df[df['unit'] == unit]

        for i in range(len(data) - SEQ_LEN):
            X.append(data[features].iloc[i:i+SEQ_LEN].values)
            y.append(data['RUL'].iloc[i+SEQ_LEN])

    return np.array(X), np.array(y)

def create_test_sequences(test, rul, features):
    X, y = [], []

    for i, unit in enumerate(test['unit'].unique()):
        data = test[test['unit'] == unit]

        if len(data) >= SEQ_LEN:
            seq = data[features].iloc[-SEQ_LEN:].values
        else:
            pad = np.zeros((SEQ_LEN - len(data), len(features)))
            seq = np.vstack((pad, data[features].values))

        X.append(seq)
        y.append(rul.iloc[i, 0])

    return np.array(X), np.array(y)

# =========================
# MODELS
# =========================
class LSTMModel(nn.Module):
    def __init__(self, input_size):
        super().__init__()
        self.lstm = nn.LSTM(input_size, 100, batch_first=True)
        self.dropout = nn.Dropout(0.3)
        self.fc = nn.Linear(100, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        out = out[:, -1, :]
        out = self.dropout(out)
        return self.fc(out)

class GRUModel(nn.Module):
    def __init__(self, input_size):
        super().__init__()
        self.gru = nn.GRU(input_size, 100, batch_first=True)
        self.dropout = nn.Dropout(0.3)
        self.fc = nn.Linear(100, 1)

    def forward(self, x):
        out, _ = self.gru(x)
        out = out[:, -1, :]
        out = self.dropout(out)
        return self.fc(out)

# =========================
# TRAIN
# =========================
def train_model(model, X, y):
    dataset = TensorDataset(torch.tensor(X, dtype=torch.float32),
                            torch.tensor(y, dtype=torch.float32))

    loader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True)

    model.to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
    loss_fn = nn.MSELoss()

    for epoch in range(EPOCHS):
        model.train()
        for xb, yb in loader:
            xb, yb = xb.to(DEVICE), yb.to(DEVICE)

            optimizer.zero_grad()
            preds = model(xb).squeeze()
            loss = loss_fn(preds, yb)
            loss.backward()
            optimizer.step()

    return model

# =========================
# UNCERTAINTY
# =========================
def predict_mc(model, X):
    model.train()
    X = torch.tensor(X, dtype=torch.float32).to(DEVICE)

    preds = []
    for _ in range(MC_SAMPLES):
        preds.append(model(X).detach().cpu().numpy())

    preds = np.array(preds)
    return preds.mean(axis=0).squeeze(), preds.std(axis=0).squeeze()

# =========================
# COSTS
# =========================
def baseline_cost(rul_pred, unc):
    return np.mean(np.maximum(0, 50 - rul_pred) + 10 * unc)

def generate_context(n):
    return np.stack([
        np.random.uniform(0.5, 1.5, n),
        np.random.uniform(0.5, 1.5, n)
    ], axis=1)

def camde_cost(pred, unc):
    context = generate_context(len(pred))
    adj = pred / (context[:,0] * context[:,1])
    risk = unc * context[:,0]
    return np.mean(np.maximum(0, 40 - adj) + 15 * risk)

# =========================
# RUN EXPERIMENT
# =========================
def run(train_path, test_path, rul_path):
    train, test, rul = load_data(train_path, test_path, rul_path)
    train, test, features = preprocess(train, test)

    X_train, y_train = create_train_sequences(train, features)
    X_test, y_test = create_test_sequences(test, rul, features)

    results = {"LSTM": [], "GRU": [], "CAMDE++": []}

    for r in range(RUNS):
        set_seed(42 + r)

        for name, Model in [("LSTM", LSTMModel), ("GRU", GRUModel)]:
            model = Model(len(features))
            model = train_model(model, X_train, y_train)

            pred, unc = predict_mc(model, X_test)

            # de-normalize
            pred = pred * MAX_RUL
            y_true = y_test

            rmse = np.sqrt(mean_squared_error(y_true, pred))
            base_c = baseline_cost(pred, unc)
            camde_c = camde_cost(pred, unc)

            results[name].append((rmse, base_c))
            results["CAMDE++"].append((rmse, camde_c))

    return results

# =========================
# PRINT
# =========================
def summarize(res, name):
    print(f"\n==== {name} ====")
    for k in res:
        rmses = [x[0] for x in res[k]]
        costs = [x[1] for x in res[k]]
        print(f"{k}: RMSE={np.mean(rmses):.3f}±{np.std(rmses):.3f}, Cost={np.mean(costs):.3f}±{np.std(costs):.3f}")

# =========================
# EXECUTE
# =========================
fd001 = run("train_FD001.txt", "test_FD001.txt", "RUL_FD001.txt")
fd004 = run("train_FD004.txt", "test_FD004.txt", "RUL_FD004.txt")

summarize(fd001, "FD001 RESULTS")
summarize(fd004, "FD004 RESULTS")
```

```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
import random

# =========================
# CONFIG
# =========================
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
SEQ_LEN = 30
EPOCHS = 30
BATCH_SIZE = 64
RUNS = 3
MC_SAMPLES = 20
RUL_CAP = 125
SEED = 42

# =========================
# SEED
# =========================
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

# =========================
# LOAD DATA
# =========================
def load_cmapss(train_path, test_path, rul_path):
    cols = ['unit', 'cycle'] + [f'op{i}' for i in range(1,4)] + [f's{i}' for i in range(1,22)]

    train = pd.read_csv(train_path, sep='\s+', header=None)
    test = pd.read_csv(test_path, sep='\s+', header=None)
    rul = pd.read_csv(rul_path, header=None)

    train.columns = cols
    test.columns = cols

    # Compute RUL
    max_cycle = train.groupby('unit')['cycle'].max()
    train['RUL'] = train.apply(lambda row: max_cycle[row['unit']] - row['cycle'], axis=1)

    # Normalize
    train['RUL'] = train['RUL'].clip(upper=RUL_CAP) / RUL_CAP

    return train, test, rul

# =========================
# PREPROCESS
# =========================
def preprocess(train, test):
    features = [c for c in train.columns if c.startswith('s')]

    scaler = MinMaxScaler()
    train[features] = scaler.fit_transform(train[features])
    test[features] = scaler.transform(test[features])

    return train, test, features

# =========================
# TRAIN SEQUENCES
# =========================
def create_train_sequences(df, features):
    X, y = [], []

    for unit in df['unit'].unique():
        data = df[df['unit'] == unit]

        for i in range(len(data) - SEQ_LEN):
            X.append(data[features].iloc[i:i+SEQ_LEN].values)
            y.append(data['RUL'].iloc[i+SEQ_LEN])

    return np.array(X), np.array(y)

# =========================
# TEST SEQUENCES
# =========================
def create_test_sequences(test, rul, features):
    X, y = [], []

    for i, unit in enumerate(test['unit'].unique()):
        data = test[test['unit'] == unit]

        if len(data) >= SEQ_LEN:
            seq = data[features].iloc[-SEQ_LEN:].values
        else:
            pad = np.zeros((SEQ_LEN - len(data), len(features)))
            seq = np.vstack((pad, data[features].values))

        X.append(seq)
        y.append(rul.iloc[i, 0])

    return np.array(X), np.array(y)

# =========================
# MODELS
# =========================
class LSTMModel(nn.Module):
    def __init__(self, input_size):
        super().__init__()
        self.lstm = nn.LSTM(input_size, 100, batch_first=True)
        self.dropout = nn.Dropout(0.3)
        self.fc = nn.Linear(100, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        out = out[:, -1, :]
        out = self.dropout(out)
        return self.fc(out)

class GRUModel(nn.Module):
    def __init__(self, input_size):
        super().__init__()
        self.gru = nn.GRU(input_size, 100, batch_first=True)
        self.dropout = nn.Dropout(0.3)
        self.fc = nn.Linear(100, 1)

    def forward(self, x):
        out, _ = self.gru(x)
        out = out[:, -1, :]
        out = self.dropout(out)
        return self.fc(out)

# =========================
# TRAINING (FIXED)
# =========================
def train_model(model, X, y):
    dataset = TensorDataset(torch.tensor(X, dtype=torch.float32),
                            torch.tensor(y, dtype=torch.float32))

    loader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True)

    model.to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
    loss_fn = nn.MSELoss()

    for epoch in range(EPOCHS):
        model.train()
        for xb, yb in loader:
            xb, yb = xb.to(DEVICE), yb.to(DEVICE)

            optimizer.zero_grad()
            preds = model(xb).squeeze()
            loss = loss_fn(preds, yb)
            loss.backward()
            optimizer.step()

    return model

# =========================
# MC DROPOUT
# =========================
def predict_with_uncertainty(model, X):
    model.train()  # keep dropout active
    X_t = torch.tensor(X, dtype=torch.float32).to(DEVICE)

    preds = []
    for _ in range(MC_SAMPLES):
        preds.append(model(X_t).detach().cpu().numpy())

    preds = np.array(preds)
    return preds.mean(axis=0).squeeze(), preds.std(axis=0).squeeze()

# =========================
# COSTS (UNCHANGED CAMDE++)
# =========================
def baseline_cost(rul_pred, unc):
    return np.mean(np.maximum(0, 50 - rul_pred) + 10 * unc)

def generate_context(n):
    return np.stack([
        np.random.uniform(0.5, 1.5, n),
        np.random.uniform(0.5, 1.5, n)
    ], axis=1)

def camde_adjustment(rul_pred, uncertainty, context):
    criticality = context[:, 0]
    load = context[:, 1]

    adjusted_rul = rul_pred / (criticality * load)
    risk = uncertainty * criticality

    return adjusted_rul, risk

def camde_cost(adjusted_rul, risk):
    return np.mean(np.maximum(0, 40 - adjusted_rul) + 15 * risk)

# =========================
# PIPELINE
# =========================
def run_experiment(train_path, test_path, rul_path):
    set_seed(SEED)

    train, test, rul = load_cmapss(train_path, test_path, rul_path)
    train, test, features = preprocess(train, test)

    X_train, y_train = create_train_sequences(train, features)
    X_test, y_test = create_test_sequences(test, rul, features)

    results = {"LSTM": [], "GRU": [], "CAMDE++": []}

    for run in range(RUNS):
        print(f"Run {run+1}")

        for name, Model in [("LSTM", LSTMModel), ("GRU", GRUModel)]:
            set_seed(SEED + run)

            model = Model(len(features))
            model = train_model(model, X_train, y_train)

            preds, unc = predict_with_uncertainty(model, X_test)

            # De-normalize
            preds = preds * RUL_CAP
            y_true = y_test

            rmse = np.sqrt(mean_squared_error(y_true, preds))
            base_c = baseline_cost(preds, unc)

            results[name].append((rmse, base_c))

            # CAMDE++ (unchanged)
            context = generate_context(len(preds))
            adj_rul, risk = camde_adjustment(preds, unc, context)
            c_cost = camde_cost(adj_rul, risk)

            results["CAMDE++"].append((rmse, c_cost))

    return results

# =========================
# SUMMARY
# =========================
def summarize(results, name):
    print(f"\n==== {name} ====")
    for model in results:
        rmses = [r[0] for r in results[model]]
        costs = [r[1] for r in results[model]]

        print(f"{model}: RMSE={np.mean(rmses):.3f}±{np.std(rmses):.3f}, "
              f"Cost={np.mean(costs):.3f}±{np.std(costs):.3f}")

# =========================
# RUN
# =========================
fd001 = run_experiment("train_FD001.txt", "test_FD001.txt", "RUL_FD001.txt")
fd004 = run_experiment("train_FD004.txt", "test_FD004.txt", "RUL_FD004.txt")

summarize(fd001, "FD001 RESULTS")
summarize(fd004, "FD004 RESULTS")
```
