```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input

# ===============================
# CONFIG
# ===============================
SEQ_LEN = 30
MAX_RUL = 130
EPOCHS = 20
BATCH_SIZE = 128

# ===============================
# LOAD DATA
# ===============================
def load_cmapss(path):
    df = pd.read_csv(path, sep=" ", header=None)
    df = df.dropna(axis=1)

    cols = ['unit','cycle'] + [f'op{i}' for i in range(1,4)] + [f's{i}' for i in range(1,22)]
    df.columns = cols

    return df.sort_values(['unit','cycle'])

# ===============================
# COMPUTE RUL (TRAIN)
# ===============================
def compute_rul_train(df):
    max_cycle = df.groupby('unit')['cycle'].max().reset_index()
    max_cycle.columns = ['unit','max_cycle']

    df = df.merge(max_cycle, on='unit')
    df['RUL'] = df['max_cycle'] - df['cycle']
    df['RUL'] = df['RUL'].clip(upper=MAX_RUL)

    return df

# ===============================
# PREPROCESS
# ===============================
def preprocess(train_df, test_df):
    scaler = MinMaxScaler()

    features = train_df.columns.difference(['unit','cycle','RUL','max_cycle'])

    train_df[features] = scaler.fit_transform(train_df[features])
    test_df[features] = scaler.transform(test_df[features])

    return train_df, test_df, features

# ===============================
# CREATE SEQUENCES
# ===============================
def create_sequences(df, features):
    X, y = [], []

    for unit in df['unit'].unique():
        data = df[df['unit'] == unit]

        for i in range(len(data) - SEQ_LEN):
            X.append(data[features].iloc[i:i+SEQ_LEN].values)
            y.append(data['RUL'].iloc[i+SEQ_LEN])

    return np.array(X), np.array(y)

# ===============================
# MODEL
# ===============================
def build_model(input_shape):
    model = Sequential([
        Input(shape=input_shape),
        LSTM(128, return_sequences=True),
        Dropout(0.3),
        LSTM(64),
        Dropout(0.3),
        Dense(32, activation='relu'),
        Dense(1)
    ])

    model.compile(optimizer='adam', loss='mse')
    return model

# ===============================
# NASA SCORE (STANDARD)
# ===============================
def nasa_score(y_true, y_pred):
    score = 0
    for i in range(len(y_true)):
        d = y_pred[i] - y_true[i]

        if d < 0:
            score += np.exp(-d / 13) - 1
        else:
            score += np.exp(d / 10) - 1

    return score / len(y_true)

# ===============================
# FAILURE PROBABILITY (FIXED)
# ===============================
def failure_prob(rul):
    return 1 / (1 + np.exp(0.1 * (rul - 40)))

# ===============================
# MAIN PIPELINE
# ===============================
def run(train_path):

    print("Loading data...")
    df = load_cmapss(train_path)

    print("Computing RUL...")
    df = compute_rul_train(df)

    # ===========================
    # TRAIN / TEST SPLIT (CRITICAL)
    # ===========================
    units = df['unit'].unique()
    split = int(len(units) * 0.8)

    train_units = units[:split]
    test_units = units[split:]

    train_df = df[df['unit'].isin(train_units)].copy()
    test_df = df[df['unit'].isin(test_units)].copy()

    print(f"Train engines: {len(train_units)}")
    print(f"Test engines: {len(test_units)}")

    # ===========================
    # PREPROCESS
    # ===========================
    train_df, test_df, features = preprocess(train_df, test_df)

    # ===========================
    # SEQUENCES
    # ===========================
    X_train, y_train = create_sequences(train_df, features)
    X_test, y_test = create_sequences(test_df, features)

    print(f"Train samples: {X_train.shape}")
    print(f"Test samples: {X_test.shape}")

    # Normalize target
    y_train = y_train / MAX_RUL
    y_test = y_test / MAX_RUL

    # ===========================
    # TRAIN MODEL
    # ===========================
    print("Training model...")
    model = build_model((X_train.shape[1], X_train.shape[2]))

    model.fit(X_train, y_train,
              validation_data=(X_test, y_test),
              epochs=EPOCHS,
              batch_size=BATCH_SIZE,
              verbose=1)

    # ===========================
    # PREDICTION
    # ===========================
    preds = model.predict(X_test)

    y_true = y_test * MAX_RUL
    y_pred = preds.flatten() * MAX_RUL

    y_pred = np.clip(y_pred, 0, MAX_RUL)

    # ===========================
    # METRICS
    # ===========================
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    nasa = nasa_score(y_true, y_pred)

    print("\n===== RESULTS =====")
    print(f"RMSE: {rmse:.2f}")
    print(f"NASA Score: {nasa:.2f}")

    # ===========================
    # BASELINE (LINEAR REGRESSION)
    # ===========================
    print("\nRunning baseline...")

    X_train_flat = X_train.reshape(X_train.shape[0], -1)
    X_test_flat = X_test.reshape(X_test.shape[0], -1)

    lr = LinearRegression()
    lr.fit(X_train_flat, y_train)

    pred_lr = lr.predict(X_test_flat) * MAX_RUL

    rmse_lr = np.sqrt(mean_squared_error(y_true, pred_lr))

    print(f"Baseline RMSE: {rmse_lr:.2f}")

    # ===========================
    # GRAPH 1
    # ===========================
    plt.figure()
    plt.scatter(y_true[:200], y_pred[:200])
    plt.xlabel("True RUL")
    plt.ylabel("Predicted RUL")
    plt.title("RUL Prediction")
    plt.show()

    # ===========================
    # GRAPH 2
    # ===========================
    plt.figure()
    errors = y_pred - y_true
    plt.hist(errors, bins=50)
    plt.title("Error Distribution")
    plt.show()

    # ===========================
    # FAILURE PROBABILITY CHECK
    # ===========================
    plt.figure()
    fp_vals = [failure_prob(r) for r in y_pred[:200]]
    plt.scatter(y_pred[:200], fp_vals)
    plt.xlabel("RUL")
    plt.ylabel("Failure Probability")
    plt.title("FP vs RUL")
    plt.show()


# ===============================
# RUN
# ===============================
if __name__ == "__main__":
    run("train_FD001.txt")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input

# ===============================
# CONFIG
# ===============================
SEQ_LEN = 30
MAX_RUL = 130
EPOCHS = 20
BATCH_SIZE = 128
MC_RUNS = 30

# ===============================
# LOAD DATA
# ===============================
def load_data(path):
    df = pd.read_csv(path, sep=" ", header=None)
    df = df.dropna(axis=1)

    cols = ['unit','cycle'] + [f'op{i}' for i in range(1,4)] + [f's{i}' for i in range(1,22)]
    df.columns = cols

    return df.sort_values(['unit','cycle'])

# ===============================
# COMPUTE RUL (TRAIN)
# ===============================
def compute_rul_train(df):
    max_cycle = df.groupby('unit')['cycle'].max().reset_index()
    max_cycle.columns = ['unit','max_cycle']

    df = df.merge(max_cycle, on='unit')
    df['RUL'] = df['max_cycle'] - df['cycle']
    df['RUL'] = df['RUL'].clip(upper=MAX_RUL)

    return df

# ===============================
# COMPUTE RUL (TEST)
# ===============================
def compute_rul_test(df, rul_file):
    rul = pd.read_csv(rul_file, header=None)
    rul.columns = ['RUL']

    max_cycle = df.groupby('unit')['cycle'].max().reset_index()
    max_cycle.columns = ['unit','max_cycle']

    max_cycle['RUL'] = rul['RUL']

    df = df.merge(max_cycle, on='unit')

    df['RUL'] = df['RUL'] + (df['max_cycle'] - df['cycle'])
    df['RUL'] = df['RUL'].clip(upper=MAX_RUL)

    return df

# ===============================
# PREPROCESS
# ===============================
def preprocess(train_df, test_df):
    scaler = MinMaxScaler()

    features = train_df.columns.difference(['unit','cycle','RUL','max_cycle'])

    train_df[features] = scaler.fit_transform(train_df[features])
    test_df[features] = scaler.transform(test_df[features])

    return train_df, test_df, features

# ===============================
# SEQUENCE CREATION
# ===============================
def create_sequences(df, features):
    X, y = [], []

    for unit in df['unit'].unique():
        data = df[df['unit'] == unit]

        for i in range(len(data) - SEQ_LEN):
            X.append(data[features].iloc[i:i+SEQ_LEN].values)
            y.append(data['RUL'].iloc[i+SEQ_LEN])

    return np.array(X), np.array(y)

# ===============================
# MODEL
# ===============================
def build_model(input_shape):
    model = Sequential([
        Input(shape=input_shape),
        LSTM(128, return_sequences=True),
        Dropout(0.3),
        LSTM(64),
        Dropout(0.3),
        Dense(32, activation='relu'),
        Dense(1)
    ])

    model.compile(optimizer='adam', loss='mse')
    return model

# ===============================
# MC DROPOUT
# ===============================
def mc_dropout(model, X, runs=30):
    preds = []
    for _ in range(runs):
        preds.append(model(X, training=True).numpy())

    preds = np.array(preds)
    return preds.mean(axis=0), preds.std(axis=0)

# ===============================
# FAILURE PROBABILITY (CALIBRATED)
# ===============================
def failure_prob(rul):
    return 1 / (1 + np.exp(0.08 * (rul - 40)))

# ===============================
# NASA SCORE
# ===============================
def nasa_score(y_true, y_pred):
    score = 0
    for i in range(len(y_true)):
        d = y_pred[i] - y_true[i]

        if d < 0:
            score += np.exp(-d / 13) - 1
        else:
            score += np.exp(d / 10) - 1

    return score / len(y_true)

# ===============================
# CONTEXT
# ===============================
def get_context(name):
    return {
        "normal":[0.5,0.5,0.5],
        "emergency":[0.9,0.2,1.0],
        "low_budget":[0.3,0.9,0.3]
    }[name]

# ===============================
# DECISION SCORE (CAMDE++)
# ===============================
def compute_score(rul, fp, u, c):
    c_risk, c_cost, c_urgency = c

    return (
        0.4*(1/(rul+1)) +
        0.3*fp +
        0.2*u +
        0.3*c_risk +
        0.3*c_urgency -
        0.3*c_cost
    )

def decision(score):
    if score < 0.35:
        return "Monitor"
    elif score < 0.65:
        return "Scheduled"
    else:
        return "Immediate"

# ===============================
# MAIN PIPELINE
# ===============================
def run(train_path, test_path, rul_path):

    print("Loading data...")
    train_df = load_data(train_path)
    test_df = load_data(test_path)

    print("Computing RUL...")
    train_df = compute_rul_train(train_df)
    test_df = compute_rul_test(test_df, rul_path)

    print("Preprocessing...")
    train_df, test_df, features = preprocess(train_df, test_df)

    print("Creating sequences...")
    X_train, y_train = create_sequences(train_df, features)
    X_test, y_test = create_sequences(test_df, features)

    # Normalize
    y_train = y_train / MAX_RUL
    y_test = y_test / MAX_RUL

    print("Training model...")
    model = build_model((X_train.shape[1], X_train.shape[2]))

    model.fit(X_train, y_train,
              validation_split=0.1,
              epochs=EPOCHS,
              batch_size=BATCH_SIZE)

    # ===========================
    # PREDICTION + UNCERTAINTY
    # ===========================
    mean, std = mc_dropout(model, X_test, MC_RUNS)

    y_pred = mean.flatten() * MAX_RUL
    y_true = y_test * MAX_RUL

    y_pred = np.clip(y_pred, 0, MAX_RUL)

    # ===========================
    # METRICS
    # ===========================
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    nasa = nasa_score(y_true, y_pred)

    print("\n===== FINAL RESULTS =====")
    print(f"RMSE: {rmse:.2f}")
    print(f"NASA Score: {nasa:.2f}")

    # ===========================
    # BASELINE
    # ===========================
    print("\nBaseline...")

    X_train_flat = X_train.reshape(X_train.shape[0], -1)
    X_test_flat = X_test.reshape(X_test.shape[0], -1)

    lr = LinearRegression()
    lr.fit(X_train_flat, y_train)

    pred_lr = lr.predict(X_test_flat) * MAX_RUL
    rmse_lr = np.sqrt(mean_squared_error(y_true, pred_lr))

    print(f"Baseline RMSE: {rmse_lr:.2f}")

    # ===========================
    # DECISION SYSTEM
    # ===========================
    print("\nEngine | Context | RUL | FP | U | Decision")

    for i in range(10):
        rul = float(y_pred[i])
        u = float(std[i][0])
        fp = failure_prob(rul)

        for ctx_name in ["normal","emergency","low_budget"]:
            ctx = get_context(ctx_name)
            score = compute_score(rul, fp, u, ctx)
            dec = decision(score)

            print(f"{i+1:<6} | {ctx_name:<10} | {rul:6.2f} | {fp:5.3f} | {u:5.3f} | {dec}")

    # ===========================
    # PLOTS
    # ===========================
    plt.figure()
    plt.scatter(y_true[:200], y_pred[:200])
    plt.xlabel("True RUL")
    plt.ylabel("Predicted RUL")
    plt.title("RUL Prediction")
    plt.show()

    plt.figure()
    errors = y_pred - y_true
    plt.hist(errors, bins=50)
    plt.title("Error Distribution")
    plt.show()


# ===============================
# RUN
# ===============================
if __name__ == "__main__":
    run("train_FD001.txt", "test_FD001.txt", "RUL_FD001.txt")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input

# ===============================
# CONFIG
# ===============================
SEQ_LEN = 30
MAX_RUL = 130
EPOCHS = 20
BATCH_SIZE = 128
MC_RUNS = 30

# ===============================
# LOAD DATA
# ===============================
def load_data(path):
    df = pd.read_csv(path, sep=" ", header=None)
    df = df.dropna(axis=1)

    cols = ['unit','cycle'] + [f'op{i}' for i in range(1,4)] + [f's{i}' for i in range(1,22)]
    df.columns = cols

    return df.sort_values(['unit','cycle'])

# ===============================
# RUL (TRAIN)
# ===============================
def compute_rul_train(df):
    max_cycle = df.groupby('unit')['cycle'].max().reset_index()
    max_cycle.columns = ['unit','max_cycle']

    df = df.merge(max_cycle, on='unit')
    df['RUL'] = df['max_cycle'] - df['cycle']
    df['RUL'] = df['RUL'].clip(upper=MAX_RUL)

    return df

# ===============================
# RUL (TEST - OFFICIAL)
# ===============================
def compute_rul_test(df, rul_file):
    rul = pd.read_csv(rul_file, header=None)
    rul.columns = ['RUL']

    max_cycle = df.groupby('unit')['cycle'].max().reset_index()
    max_cycle.columns = ['unit','max_cycle']

    max_cycle['RUL'] = rul['RUL']

    df = df.merge(max_cycle, on='unit')
    df['RUL'] = df['RUL'] + (df['max_cycle'] - df['cycle'])
    df['RUL'] = df['RUL'].clip(upper=MAX_RUL)

    return df

# ===============================
# PREPROCESS
# ===============================
def preprocess(train_df, test_df):
    scaler = MinMaxScaler()

    features = train_df.columns.difference(['unit','cycle','RUL','max_cycle'])

    train_df[features] = scaler.fit_transform(train_df[features])
    test_df[features] = scaler.transform(test_df[features])

    return train_df, test_df, features

# ===============================
# TRAIN SEQUENCES (UNCHANGED)
# ===============================
def create_sequences(df, features):
    X, y = [], []

    for unit in df['unit'].unique():
        data = df[df['unit'] == unit]

        for i in range(len(data) - SEQ_LEN):
            X.append(data[features].iloc[i:i+SEQ_LEN].values)
            y.append(data['RUL'].iloc[i+SEQ_LEN])

    return np.array(X), np.array(y)

# ===============================
# TEST SEQUENCES (FIXED — CRITICAL)
# ===============================
def create_last_sequences(df, features):
    X, y = [], []

    for unit in df['unit'].unique():
        data = df[df['unit'] == unit]

        if len(data) >= SEQ_LEN:
            seq = data[features].iloc[-SEQ_LEN:].values
            target = data['RUL'].iloc[-1]

            X.append(seq)
            y.append(target)

    return np.array(X), np.array(y)

# ===============================
# MODEL
# ===============================
def build_model(input_shape):
    model = Sequential([
        Input(shape=input_shape),
        LSTM(128, return_sequences=True),
        Dropout(0.3),
        LSTM(64),
        Dropout(0.3),
        Dense(32, activation='relu'),
        Dense(1)
    ])

    model.compile(optimizer='adam', loss='mse')
    return model

# ===============================
# MC DROPOUT (UNCHANGED)
# ===============================
def mc_dropout(model, X, runs=30):
    preds = []
    for _ in range(runs):
        preds.append(model(X, training=True).numpy())

    preds = np.array(preds)
    return preds.mean(axis=0), preds.std(axis=0)

# ===============================
# FAILURE PROBABILITY
# ===============================
def failure_prob(rul):
    return 1 / (1 + np.exp(0.08 * (rul - 40)))

# ===============================
# NASA SCORE (CORRECT)
# ===============================
def nasa_score(y_true, y_pred):
    score = 0
    for i in range(len(y_true)):
        d = y_pred[i] - y_true[i]

        if d < 0:
            score += np.exp(-d / 13) - 1
        else:
            score += np.exp(d / 10) - 1

    return score / len(y_true)

# ===============================
# CONTEXT + DECISION (UNCHANGED)
# ===============================
def get_context(name):
    return {
        "normal":[0.5,0.5,0.5],
        "emergency":[0.9,0.2,1.0],
        "low_budget":[0.3,0.9,0.3]
    }[name]

def compute_score(rul, fp, u, c):
    c_risk, c_cost, c_urgency = c

    return (
        0.4*(1/(rul+1)) +
        0.3*fp +
        0.2*u +
        0.3*c_risk +
        0.3*c_urgency -
        0.3*c_cost
    )

def decision(score):
    if score < 0.35:
        return "Monitor"
    elif score < 0.65:
        return "Scheduled"
    else:
        return "Immediate"

# ===============================
# MAIN PIPELINE
# ===============================
def run(train_path, test_path, rul_path):

    print("Loading data...")
    train_df = load_data(train_path)
    test_df = load_data(test_path)

    print("Computing RUL...")
    train_df = compute_rul_train(train_df)
    test_df = compute_rul_test(test_df, rul_path)

    print("Preprocessing...")
    train_df, test_df, features = preprocess(train_df, test_df)

    print("Creating sequences...")
    X_train, y_train = create_sequences(train_df, features)
    X_test, y_test = create_last_sequences(test_df, features)

    # Normalize
    y_train = y_train / MAX_RUL
    y_test = y_test / MAX_RUL

    print(f"Train samples: {X_train.shape}")
    print(f"Test engines (correct eval): {X_test.shape[0]}")

    print("Training model...")
    model = build_model((X_train.shape[1], X_train.shape[2]))

    model.fit(X_train, y_train,
              validation_split=0.1,
              epochs=EPOCHS,
              batch_size=BATCH_SIZE)

    # ===========================
    # PREDICTION + UNCERTAINTY
    # ===========================
    mean, std = mc_dropout(model, X_test, MC_RUNS)

    y_pred = mean.flatten() * MAX_RUL
    y_true = y_test * MAX_RUL

    y_pred = np.clip(y_pred, 0, MAX_RUL)

    # ===========================
    # METRICS
    # ===========================
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    nasa = nasa_score(y_true, y_pred)

    print("\n===== FINAL RESULTS =====")
    print(f"RMSE: {rmse:.2f}")
    print(f"NASA Score: {nasa:.2f}")

    # ===========================
    # BASELINE
    # ===========================
    X_train_flat = X_train.reshape(X_train.shape[0], -1)
    X_test_flat = X_test.reshape(X_test.shape[0], -1)

    lr = LinearRegression()
    lr.fit(X_train_flat, y_train)

    pred_lr = lr.predict(X_test_flat) * MAX_RUL
    rmse_lr = np.sqrt(mean_squared_error(y_true, pred_lr))

    print(f"Baseline RMSE: {rmse_lr:.2f}")

    # ===========================
    # DECISION SYSTEM
    # ===========================
    print("\nEngine | Context | RUL | FP | U | Decision")

    for i in range(len(y_pred)):
        rul = float(y_pred[i])
        u = float(std[i][0])
        fp = failure_prob(rul)

        for ctx_name in ["normal","emergency","low_budget"]:
            ctx = get_context(ctx_name)
            score = compute_score(rul, fp, u, ctx)
            dec = decision(score)

            print(f"{i+1:<6} | {ctx_name:<10} | {rul:6.2f} | {fp:5.3f} | {u:5.3f} | {dec}")

# ===============================
# RUN
# ===============================
if __name__ == "__main__":
    run("train_FD001.txt", "test_FD001.txt", "RUL_FD001.txt")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout, Input

# ===============================
# CONFIG
# ===============================
SEQ_LEN = 30
MAX_RUL = 130
EPOCHS = 20
BATCH_SIZE = 128
MC_RUNS = 30

# ===============================
# LOAD DATA
# ===============================
def load_data(path):
    df = pd.read_csv(path, sep=" ", header=None)
    df = df.dropna(axis=1)

    cols = ['unit','cycle'] + [f'op{i}' for i in range(1,4)] + [f's{i}' for i in range(1,22)]
    df.columns = cols

    return df.sort_values(['unit','cycle'])

# ===============================
# TRAIN RUL (CLIPPED)
# ===============================
def compute_rul_train(df):
    max_cycle = df.groupby('unit')['cycle'].max().reset_index()
    max_cycle.columns = ['unit','max_cycle']

    df = df.merge(max_cycle, on='unit')
    df['RUL'] = df['max_cycle'] - df['cycle']

    # Clip ONLY training
    df['RUL'] = df['RUL'].clip(upper=MAX_RUL)

    return df

# ===============================
# TEST RUL (UNCLIPPED — CRITICAL)
# ===============================
def compute_rul_test(df, rul_path):
    rul = pd.read_csv(rul_path, header=None)
    rul.columns = ['RUL']

    max_cycle = df.groupby('unit')['cycle'].max().reset_index()
    max_cycle.columns = ['unit','max_cycle']

    max_cycle['RUL'] = rul['RUL']

    df = df.merge(max_cycle, on='unit')

    # DO NOT CLIP — this is the fix
    df['RUL'] = df['RUL'] + (df['max_cycle'] - df['cycle'])

    return df

# ===============================
# PREPROCESS
# ===============================
def preprocess(train_df, test_df):
    scaler = MinMaxScaler()

    features = train_df.columns.difference(['unit','cycle','RUL','max_cycle'])

    train_df[features] = scaler.fit_transform(train_df[features])
    test_df[features] = scaler.transform(test_df[features])

    return train_df, test_df, features

# ===============================
# TRAIN SEQUENCES
# ===============================
def create_sequences(df, features):
    X, y = [], []

    for unit in df['unit'].unique():
        data = df[df['unit'] == unit]

        for i in range(len(data) - SEQ_LEN):
            X.append(data[features].iloc[i:i+SEQ_LEN].values)
            y.append(data['RUL'].iloc[i+SEQ_LEN])

    return np.array(X), np.array(y)

# ===============================
# TEST (LAST CYCLE ONLY)
# ===============================
def create_last_sequences(df, features):
    X, y = [], []

    for unit in df['unit'].unique():
        data = df[df['unit'] == unit]

        if len(data) >= SEQ_LEN:
            seq = data[features].iloc[-SEQ_LEN:].values
            target = data['RUL'].iloc[-1]

            X.append(seq)
            y.append(target)

    return np.array(X), np.array(y)

# ===============================
# MODEL
# ===============================
def build_model(input_shape):
    model = Sequential([
        Input(shape=input_shape),
        LSTM(128, return_sequences=True),
        Dropout(0.3),
        LSTM(64),
        Dropout(0.3),
        Dense(32, activation='relu'),
        Dense(1)
    ])

    model.compile(optimizer='adam', loss='mse')
    return model

# ===============================
# MC DROPOUT
# ===============================
def mc_dropout(model, X, runs=30):
    preds = []
    for _ in range(runs):
        preds.append(model(X, training=True).numpy())

    preds = np.array(preds)
    return preds.mean(axis=0), preds.std(axis=0)

# ===============================
# NASA SCORE (STANDARD)
# ===============================
def nasa_score(y_true, y_pred):
    score = 0
    for i in range(len(y_true)):
        d = y_pred[i] - y_true[i]

        if d < 0:
            score += np.exp(-d / 13) - 1
        else:
            score += np.exp(d / 10) - 1

    return score / len(y_true)

# ===============================
# FAILURE PROBABILITY
# ===============================
def failure_prob(rul):
    return 1 / (1 + np.exp(0.08 * (rul - 40)))

# ===============================
# DECISION SYSTEM
# ===============================
def get_context(name):
    return {
        "normal":[0.5,0.5,0.5],
        "emergency":[0.9,0.2,1.0],
        "low_budget":[0.3,0.9,0.3]
    }[name]

def compute_score(rul, fp, u, c):
    c_risk, c_cost, c_urgency = c

    return (
        0.4*(1/(rul+1)) +
        0.3*fp +
        0.2*u +
        0.3*c_risk +
        0.3*c_urgency -
        0.3*c_cost
    )

def decision(score):
    if score < 0.35:
        return "Monitor"
    elif score < 0.65:
        return "Scheduled"
    else:
        return "Immediate"

# ===============================
# MAIN
# ===============================
def run(train_path, test_path, rul_path):

    print("Loading data...")
    train_df = load_data(train_path)
    test_df = load_data(test_path)

    print("Computing RUL...")
    train_df = compute_rul_train(train_df)
    test_df = compute_rul_test(test_df, rul_path)

    print("Preprocessing...")
    train_df, test_df, features = preprocess(train_df, test_df)

    print("Creating sequences...")
    X_train, y_train = create_sequences(train_df, features)
    X_test, y_test = create_last_sequences(test_df, features)

    # Normalize training only
    y_train = y_train / MAX_RUL

    print(f"Train samples: {X_train.shape}")
    print(f"Test engines: {X_test.shape[0]}")

    print("Training model...")
    model = build_model((X_train.shape[1], X_train.shape[2]))

    model.fit(
        X_train,
        y_train,
        validation_split=0.1,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE
    )

    # ===========================
    # PREDICTION
    # ===========================
    mean, std = mc_dropout(model, X_test, MC_RUNS)

    y_pred = mean.flatten() * MAX_RUL
    y_pred = np.clip(y_pred, 0, MAX_RUL)

    y_true = y_test  # UNCLIPPED TRUE VALUES

    # ===========================
    # METRICS
    # ===========================
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    nasa = nasa_score(y_true, y_pred)

    print("\n===== FINAL RESULTS =====")
    print(f"RMSE: {rmse:.2f}")
    print(f"NASA Score: {nasa:.2f}")

    # ===========================
    # BASELINE
    # ===========================
    X_train_flat = X_train.reshape(X_train.shape[0], -1)
    X_test_flat = X_test.reshape(X_test.shape[0], -1)

    lr = LinearRegression()
    lr.fit(X_train_flat, y_train)

    pred_lr = lr.predict(X_test_flat) * MAX_RUL
    rmse_lr = np.sqrt(mean_squared_error(y_true, pred_lr))

    print(f"Baseline RMSE: {rmse_lr:.2f}")

# ===============================
# RUN
# ===============================
if __name__ == "__main__":
    run("train_FD001.txt", "test_FD001.txt", "RUL_FD001.txt")
```

```python
"""
=============================================================================
CAMDE++ — REAL IMPLEMENTATION (IEEE READY)
NASA C-MAPSS FD001 + LSTM + MC Dropout + Context-Aware Evaluation
=============================================================================
"""

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
import tensorflow as tf

# ─────────────────────────────────────────────────────────────
# 1. LOAD NASA C-MAPSS DATA (FD001)
# ─────────────────────────────────────────────────────────────

def load_data():
    train = pd.read_csv("train_FD001.txt", sep="\s+", header=None)
    test  = pd.read_csv("test_FD001.txt",  sep="\s+", header=None)
    rul   = pd.read_csv("RUL_FD001.txt",   sep="\s+", header=None)

    cols = ['engine_id', 'cycle'] + [f's{i}' for i in range(1, 22)]
    train.columns = cols
    test.columns  = cols

    return train, test, rul

train_df, test_df, rul_df = load_data()

# ─────────────────────────────────────────────────────────────
# 2. PREPROCESSING
# ─────────────────────────────────────────────────────────────

RUL_CAP = 125

def add_rul(df):
    max_cycle = df.groupby('engine_id')['cycle'].max()
    df['max_cycle'] = df['engine_id'].map(max_cycle)
    df['rul'] = df['max_cycle'] - df['cycle']
    df['rul'] = df['rul'].clip(upper=RUL_CAP)
    return df.drop(columns=['max_cycle'])

train_df = add_rul(train_df)

# Normalize
sensor_cols = [col for col in train_df.columns if col.startswith('s')]
scaler = MinMaxScaler()
train_df[sensor_cols] = scaler.fit_transform(train_df[sensor_cols])
test_df[sensor_cols]  = scaler.transform(test_df[sensor_cols])

# ─────────────────────────────────────────────────────────────
# 3. SEQUENCE GENERATION
# ─────────────────────────────────────────────────────────────

SEQ_LEN = 30

def create_sequences(df):
    X, y = [], []
    for eng in df['engine_id'].unique():
        eng_df = df[df['engine_id'] == eng]
        for i in range(len(eng_df) - SEQ_LEN):
            X.append(eng_df[sensor_cols].iloc[i:i+SEQ_LEN].values)
            y.append(eng_df['rul'].iloc[i+SEQ_LEN])
    return np.array(X), np.array(y)

X_train, y_train = create_sequences(train_df)

# Test sequences (last window per engine)
def create_test_sequences(df):
    X_test = []
    for eng in df['engine_id'].unique():
        eng_df = df[df['engine_id'] == eng]
        X_test.append(eng_df[sensor_cols].iloc[-SEQ_LEN:].values)
    return np.array(X_test)

X_test = create_test_sequences(test_df)

true_rul = rul_df[0].values
true_rul = np.clip(true_rul, 0, RUL_CAP)

# ─────────────────────────────────────────────────────────────
# 4. LSTM MODEL
# ─────────────────────────────────────────────────────────────

model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(SEQ_LEN, len(sensor_cols))),
    Dropout(0.3),
    LSTM(32),
    Dropout(0.3),
    Dense(1)
])

model.compile(optimizer='adam', loss='mse')

model.fit(X_train, y_train, epochs=15, batch_size=64, validation_split=0.2)

# ─────────────────────────────────────────────────────────────
# 5. MC DROPOUT (UNCERTAINTY)
# ─────────────────────────────────────────────────────────────

def mc_dropout_predict(model, X, T=30):
    preds = []
    for _ in range(T):
        preds.append(model(X, training=True).numpy().flatten())
    preds = np.array(preds)
    return preds.mean(axis=0), preds.std(axis=0)

pred_rul, uncertainty = mc_dropout_predict(model, X_test)

# ─────────────────────────────────────────────────────────────
# 6. METRICS
# ─────────────────────────────────────────────────────────────

def rmse(y_true, y_pred):
    return np.sqrt(mean_squared_error(y_true, y_pred))

def nasa_score(y_true, y_pred):
    d = y_pred - y_true
    return np.sum(np.where(d < 0,
                           np.exp(-d / 13) - 1,
                           np.exp(d / 10) - 1))

print("RMSE:", rmse(true_rul, pred_rul))
print("NASA Score:", nasa_score(true_rul, pred_rul))

# ─────────────────────────────────────────────────────────────
# 7. CONTEXT DERIVATION (REALISTIC)
# ─────────────────────────────────────────────────────────────

def derive_context(X_seq):
    # Use variance as proxy for stress
    var = np.var(X_seq, axis=(1,2))
    context = []
    for v in var:
        if v < 0.01:
            context.append("Low")
        elif v < 0.03:
            context.append("Medium")
        else:
            context.append("High")
    return np.array(context)

context = derive_context(X_test)

# ─────────────────────────────────────────────────────────────
# 8. DECISION SYSTEM (CAMDE++)
# ─────────────────────────────────────────────────────────────

def decision(rul, ctx):
    thresholds = {
        "Low": (50,30,10),
        "Medium": (70,45,15),
        "High": (90,60,20)
    }
    m,i,s = thresholds[ctx]

    if rul <= s: return 3
    elif rul <= i: return 2
    elif rul <= m: return 1
    else: return 0

decisions = np.array([decision(pred_rul[i], context[i]) for i in range(len(pred_rul))])

# ─────────────────────────────────────────────────────────────
# 9. COST ANALYSIS
# ─────────────────────────────────────────────────────────────

COST = {
    "continue": 0,
    "inspect": 0.15,
    "maintain": 0.5,
    "shutdown": 0.8,
    "failure": 1.5
}

def compute_cost(r_pred, r_true, ctx):
    act = decision(r_pred, ctx)
    act_map = ["continue","inspect","maintain","shutdown"]

    if r_true < 15 and act == 0:
        return COST["failure"]
    return COST[act_map[act]]

costs = np.array([
    compute_cost(pred_rul[i], true_rul[i], context[i])
    for i in range(len(pred_rul))
])

print("Total Cost:", costs.sum())

# ─────────────────────────────────────────────────────────────
# 10. VISUALIZATION
# ─────────────────────────────────────────────────────────────

plt.figure(figsize=(8,6))
plt.scatter(true_rul, pred_rul, alpha=0.6)
plt.plot([0,150],[0,150],'r--')
plt.xlabel("True RUL")
plt.ylabel("Predicted RUL")
plt.title("Prediction vs True")
plt.show()

# Uncertainty plot
plt.figure(figsize=(8,6))
plt.scatter(pred_rul, uncertainty, alpha=0.6)
plt.xlabel("Predicted RUL")
plt.ylabel("Uncertainty")
plt.title("Uncertainty Distribution")
plt.show()

# Cost distribution
plt.figure(figsize=(8,6))
plt.hist(costs, bins=20)
plt.title("Cost Distribution")
plt.show()
```
